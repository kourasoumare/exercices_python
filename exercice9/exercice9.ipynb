```python
import pandas as pd
titanic = pd.read_csv("../data/titanic_propre.csv")
```

```python
titanic.groupby(["class", "sex"])["survived"].mean()
```

Oui dans chaque classe les femmes ont beaucoup mieux survécu que les hommes. En 1ere classe 96,7 % des femmes ont survécu contre 36,9 % des hommes. En 2e classe 92,1 % contre 15,7 %, c’est là que l’écart est le plus grand. En 3e classe 50 % contre 13,5 %. Les femmes ont sûrement été prioritaires lors du sauvetage

```python
titanic["tranche_age"] = pd.cut(
    titanic["age"],
    bins=[0, 12, 18, 60, 100],
    labels=["enfant", "ado", "adulte", "senior"],
)
```

```python
titanic["tranche_age"].value_counts()
```

```python
titanic.groupby("tranche_age")["survived"].mean()
```

Oui les enfants ont été prioritaires. Leur taux de survie est de 58 % contre 36,5 % pour les adultes. Plus les passagers sont jeunes plus ils ont survécu.les 177 âges que j’ai remplis à l’exercice 8 sont tous tombés dans la tranche adulte ce qui peut un peu fausser les résultats 

```python
titanic.groupby("embark_town")["fare"].mean()
```

```python
pd.crosstab(titanic["embark_town"], titanic["class"], normalize="index")
```

C’est à Cherbourg que le prix moyen du billet est le plus élevé (59,95), contre 27,08 à Southampton et 13,28 à Queenstown. La répartition des classes l’explique : à Cherbourg, la moitié des passagers (50,6 %) voyageaient en 1re classe, alors qu’à Queenstown, presque tous (93,5 %) étaient en 3e classe. Ce n’est donc pas le port qui rend le billet plus cher, mais le type de passagers qui y sont montés
