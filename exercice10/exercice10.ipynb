```python
import pandas as pd
titanic = pd.read_csv("../data/titanic_propre.csv")
titanic.shape
```

```python
ports = pd.DataFrame({
    "embark_town": ["Southampton", "Cherbourg", "Queenstown"],
    "pays": ["Angleterre", "France", "Irlande"],
})
ports
```

```python
nb_avant = len(titanic)
nb_avant
```

```python
titanic = titanic.merge(ports, on="embark_town", how="left", validate="many_to_one")
titanic.head()
```

```python
assert len(titanic) == nb_avant
len(titanic)
```

J’ai joint la table ports à titanic avec how="left" pour garder tous les passagers. J’ai compté les lignes avant et après la jointure et il y en a 889 dans les deux cas

```python
titanic["pays"].value_counts()
```

```python
titanic.groupby("pays")["survived"].mean()
```

La majorité des passagers sont partis d’Angleterre (644), puis de France (168) et d’Irlande (77). Les passagers partis de France ont le mieux survécu (55,4 %), sûrement parce que la moitié d’entre eux voyageaient en 1ere classe 

```python
ports_sans_queenstown = ports[ports["embark_town"] != "Queenstown"]
ports_sans_queenstown
```

```python
titanic_base = pd.read_csv("../data/titanic_propre.csv")
```

```python
jointure_inner = titanic_base.merge(ports_sans_queenstown, on="embark_town", how="inner")
len(jointure_inner)
```

```python
jointure_left = titanic_base.merge(ports_sans_queenstown, on="embark_town", how="left")
len(jointure_left)
```

```python
jointure_left["pays"].isna().sum()
```

J’ai retiré Queenstown de la table ports puis j’ai refait la jointure de deux façons.
Avec how="inner" il ne reste que 812 passagers, les 77 passagers partis de Queenstown ont disparu car leur port n’existe plus dans ports.Avec how="left" les 889 passagers sont toujours là. Les 77 passagers de Queenstown sont gardés, mais leur colonne pays est vide.Inner peut faire perdre des lignes sans prévenir alors que left les garde et montre clairement ce qui manque.
