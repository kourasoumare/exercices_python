```python
import pandas as pd
import seaborn as sns

taxis = sns.load_dataset("taxis")
taxis.shape
```

```python
taxis.head()
```

```python
taxis.dtypes
```

```python
taxis["pickup"] = pd.to_datetime(taxis["pickup"])
taxis["dropoff"] = pd.to_datetime(taxis["dropoff"])
taxis.dtypes
```

Dans ma version de seaborn pickup et dropoff sont déjà chargées comme des dates. Je les convertis quand même avec pd.to_datetime pour que le notebook fonctionne aussi si elles arrivent sous forme de texte

```python
taxis["pickup"].min(), taxis["pickup"].max()
```

Le jeu de données couvre le mois de mars 2019 du 28 février à 23h29 au 31 mars à 23h43.

```python
ca_par_jour = taxis.set_index("pickup").resample("D")["total"].sum()
ca_par_jour
```

```python
courses_par_jour = taxis.set_index("pickup").resample("D")["total"].count()
courses_par_jour
```

en mars 2019, il y a entre 169 et 260 courses par jour pour un chiffre d’affaires d’environ 3 000$ à 4 900$  par jour. Le 28 février n’a qu’une seule course (6,30$) car le jeu de données commence à 23h29 ce jour-là.

```python
taxis["heure"] = taxis["pickup"].dt.hour
taxis[["pickup", "heure"]].head()
```

```python
courses_par_heure = taxis["heure"].value_counts()
courses_par_heure
```

L’heure la plus chargée est 18h (417 courses), suivie de 19h et 17h correspondant aux heures de sortie du travail. Les heures les plus calmes sont entre 3h et 5h du matin 

```python
taxis["jour_semaine"] = taxis["pickup"].dt.day_name()
taxis[["pickup", "jour_semaine"]].head()
```

```python
pourboir_par_jour = taxis.groupby("jour_semaine")["tip"].mean().sort_values(ascending=False)
pourboir_par_jour
```

C’est le mercredi que le pourboire moyen est le plus élevé 2,09 $, et le samedi qu’il est le plus faible 1,82 $.

```python
#calcule de la durée des courses
taxis["duree"] = taxis["dropoff"] - taxis["pickup"]
taxis["duree"].head()
```

```python
#transformation de la durée en minutes
taxis["duree_min"] = taxis["duree"].dt.total_seconds() / 60
taxis["duree_min"].head()
```

```python
#voir le resumé des durées 
taxis["duree_min"].describe()
```

```python
#verifie si les courses de moins de une minutes sont des erreurs
courses_tres_courtes = taxis[taxis["duree_min"] < 1]
print(len(courses_tres_courtes))
courses_tres_courtes[["pickup", "dropoff", "distance", "total"]].head(10)
```

```python
#veriffie la distance des courses de plus d'une heure
taxis.sort_values("duree_min", ascending=False)[["duree_min", "distance", "pickup_zone", "dropoff_zone"]].head(5)
```

```python
nb_avant = len(taxis)
taxis = taxis[taxis["duree_min"] >= 1]
print("avant:", nb_avant, "après:", len(taxis))
```
