```python
import seaborn as sns
titanic = sns.load_dataset("titanic")
```

```python
titanic["survived"].value_counts().reindex([1, 0])
```

```python
titanic["survived"].unique()
```

```python
mineur = (titanic["age"] < 18).sum()
print(f"Nombre de mineurs à bord : {mineur}")
```

```python
print("les 5 billets les plus chers :")
billets_chers= titanic.sort_values("fare", ascending=False)[["fare", "class", "embark_town"]].head(5)
print(billets_chers)
```

```python
femme_en_First=titanic[(titanic["sex"]=="female") & (titanic["class"]=="First")].shape[0]
print(femme_en_First)
personne_en_first=(titanic["class"]=="First").sum()
print(personne_en_first)
pourcentage=femme_en_First/personne_en_first*100
print(f"le pourcentage de femmes de 1ere classe est de : {pourcentage:.2f} %")
```

```python
femme_survecue_en_first= titanic[(titanic["sex"] == "female") & (titanic["class"] == "First") & (titanic["survived"] == 1)].shape[0]
femme_en_First=titanic[(titanic["sex"]=="female") & (titanic["class"]=="First")].shape[0]
print(femme_en_First)
taux_femme_survecu_first=femme_survecue_en_first/femme_en_First*100
print(f"Survie femmes 1ere classe : {taux_femme_survecu_first:.2F}")

homme_survecue_en_third= titanic[(titanic["sex"] == "male") & (titanic["class"] == "Third") & (titanic["survived"] == 1)].shape[0]
homme_en_third=titanic[(titanic["sex"]=="male") & (titanic["class"]=="Third")].shape[0]
print(homme_en_third)
taux_hommes_survecu_third=homme_survecue_en_third/homme_en_third*100
print(f"Survie hommes 3eme classe : {taux_hommes_survecu_third:.2F}")
```
